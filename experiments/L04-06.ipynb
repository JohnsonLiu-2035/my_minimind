```python
import torch
import torch.nn as nn

V, d = 6400, 768
W = nn.Embedding(V, d)

# 方法1：查表（实际实现）
token_id = 356
embedding = W(torch.tensor(token_id))  # 直接取第356行

# 方法2：矩阵乘法（数学等价，但更慢）
one_hot = torch.zeros(V)
one_hot[356] = 1.0
embedding_v2 = one_hot @ W.weight  # 矩阵乘法

print(embedding, embedding_v2)
# embedding ≈ embedding_v2（结果相同）
```

```python
# 输入：一批 token id
input_ids = torch.tensor([[356, 1892, 478, 23]])  # shape: (1, 4)

# 通过 Embedding 层
embed = nn.Embedding(6400, 768)
print(embed)
print(embed.weight)
output = embed(input_ids)  # 在embed.weight表中找到tokenid为input_ids的token
# embed 是一个对象变量，不是普通函数；只是 nn.Module 让这个对象拥有了“像函数一样被调用”的能力。
# output 是一个batch为1，token为4，每个token是一个有768维度向量的张量
print(output.shape)
# 每个 token id 变成了一个 768 维向量
```
