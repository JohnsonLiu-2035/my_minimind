```python
w = torch.tensor(3.0, requires_grad=True)
y = w ** 2 + 2 * w + 1
y.backward() # 计算梯度（求导数），保存在 w.grad 中
print(w.grad)  # 输出梯度值，即 y 对 w 的导数 dy/dw = 2*w + 2
```

```python
w = torch.tensor([3.0,4.0], requires_grad=True)
y = w ** 2 + 2 * w + 1

loss = y.sum() # 算张量中各自的梯度
loss.backward() # 计算梯度（求导数），保存在 w.grad 中
                # w.grad 的形状与 w 一样，每个位置保存目标对对应元素的偏导数
print(w.grad)  # 输出梯度值，即 y 对 w 的导数 dy/dw = 2*w + 2
```

```python
w = torch.tensor([3.0,4.0], requires_grad=True)
y = w ** 2 + 2 * w + 1

y.backward(torch.ones_like(y)) # 创建与 y 形状相同的全 1 张量：[1., 1.]
                               # 给backwards传1:1的权重  
                               # 也可以y.backward(torch.tensor([1., 1.]))                             
print(w.grad)  # 输出梯度值，即 y 对 w 的导数 dy/dw = 2*w + 2
```

```python
import torch
import torch.nn as nn

class SimpleLayer(nn.Module):  # 继承 PyTorch 神经网络模块
    def __init__(self, input_size, output_size): # 构造方法，也是一种函数
        super().__init__()    # 初始化父类
        self.linear = nn.Linear(input_size, output_size)

    def forward(self, x):     # 定义输入如何变成输出
        return self.linear(x) # linear只处理最后一维，其他维度保持不变

layer = SimpleLayer(768, 256)  # 创建层
x = torch.randn(2, 768)       # 2 个样本，每个有 768 个特征
output = layer(x)            # 自动执行 forward
print(output.shape)         # torch.Size([2, 256])
```

```python
import torch
import torch.nn as nn

# === 1. 准备数据 ===
text = "hello world hello pytorch hello deep learning"
chars = sorted(set(text))            # 所有不重复的字符
vocab_size = len(chars)               # 词表大小
char_to_id = {c: i for i, c in enumerate(chars)} # 生成{字母:索引}的字典
id_to_char = {i: c for i, c in enumerate(chars)} # 生成{索引:字母}的字典

data = torch.tensor([char_to_id[c] for c in text]) # 原始输入数据中的各个字母（空格）在字符表中的编号

# === 2. 定义模型 ===
class TinyLM(nn.Module):
    def __init__(self, vocab_size, hidden_size=32):  # hidden_size是每个字符的向量参数量，可以自己选择
        super().__init__()
        self.embed = nn.Embedding(vocab_size, hidden_size)
        self.linear1 = nn.Linear(hidden_size, hidden_size) # 同维度乘一个权重矩阵
        self.linear2 = nn.Linear(hidden_size, vocab_size) #？

    def forward(self, x):
        h = self.embed(x)               # token_id → 向量
        h = torch.relu(self.linear1(h))  # 非线性变换(激活)，让神经网络表达复杂关系。
        logits = self.linear2(h)         # 向量 → 词表大小的分数
        return logits # logits : 未归一化类别分数

# === 3. 训练 ===
model = TinyLM(vocab_size)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(500):
    inputs = data[:-1]     # "hello world ... learnin"
    targets = data[1:]     # "ello world ... learning"

    logits = model(inputs)
    loss = loss_fn(logits, targets) #这个时候，data已经是索引集了

    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    if (epoch + 1) % 50 == 0: # 每50次打印一次成果
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# === 4. 生成 ===
with torch.no_grad(): # with 用处是搭建一个条件环境，离开这个缩进后，该条件自动消失
    seed = char_to_id['h']  # 给模型一个起点
    result = ['h']
    current = torch.tensor([seed]) 

    for _ in range(4):
        logits = model(current)
        probs = torch.softmax(logits, dim=-1) # dim=-1表示对最后一维,也就是有候选字符做softmax概率分布
        next_id = int(torch.multinomial(probs, 1).item()) # multinomial是按照分布随机采样下一个输出字符
        result.append(id_to_char[next_id])
        current = torch.tensor([next_id])

    print("生成结果:", ''.join(result))
```
